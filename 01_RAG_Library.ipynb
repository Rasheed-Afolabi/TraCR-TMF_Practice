###Read the MITRE data and examine technique

```python
import json

# Locate the MITRE file we downloaded.
source_file = "/content/drive/MyDrive/TraCR_RAG_Fresh/enterprise-attack-17.1.json"

# Read the saved file.
with open(source_file, "r", encoding="utf-8") as mitre_file:
    attack_data = json.load(mitre_file)

# Choose 10 techniques for our first practice batch.
technique_ids = [
    "T1001", "T1005", "T1020", "T1021", "T1027",
    "T1033", "T1036", "T1040", "T1041", "T1046"
]

techniques_shown = 0

for record in attack_data["objects"]:
    # Skip records that are not attack techniques.
    if record["type"] != "attack-pattern":
        continue

    # Find the technique's official MITRE ID.
    for reference in record["external_references"]:
        if reference["source_name"] == "mitre-attack":
            technique_id = reference["external_id"]

            # Display the technique if it is in our chosen list.
            if technique_id in technique_ids:
                print("\n------------------------------")
                print(technique_id, "-", record["name"])
                print(record["description"])

                techniques_shown = techniques_shown + 1
                break

# Confirm how many techniques were displayed.
print("\nTotal techniques displayed:", techniques_shown)
```

### Create a reference file for each MITRE technique

The ATT&CK file holds every technique in one large record. This step splits it into separate files,
one per technique, each named by its ID (T1001.txt, T1005.txt, and so on) and saved in the data folder.
Only the 62 techniques used in this study are written.

```python
import os

# The 62 technique IDs used in this study.
labels = ['T1495','T1485','T1595','T1134','T1040','T1132','T1098','T1069','T1036','T1562',
          'T1187','T1486','T1119','T1027','T1498','T1654','T1548','T1082','T1552','T1614',
          'T1531','T1204','T1529','T1046','T1489','T1195','T1566','T1659','T1059','T1213',
          'T1133','T1080','T1005','T1078','T1001','T1190','T1203','T1136','T1491','T1033',
          'T1189','T1068','T1652','T1049','T1020','T1041','T1021','T1105','T1518','T1200',
          'T1053','T1557','T1056','T1087','T1565','T1499','T1657','T1559','T1074','T1106',
          'T1560','T1589']

# The folder that will hold one text file per technique.
data_folder = "/content/drive/MyDrive/TraCR_RAG_Fresh/data"

# Create the folder if it is not there yet.
os.makedirs(data_folder, exist_ok=True)

files_written = 0

for record in attack_data["objects"]:
    # Skip records that are not attack techniques.
    if record["type"] != "attack-pattern":
        continue

    # Find the technique's official MITRE ID.
    for reference in record["external_references"]:
        if reference["source_name"] == "mitre-attack":
            technique_id = reference["external_id"]

            # Write the technique to its own file if it is one we need.
            if technique_id in labels:
                # The file name must be the ID alone, such as T1001.txt.
                file_path = os.path.join(data_folder, technique_id + ".txt")

                with open(file_path, "w", encoding="utf-8") as technique_file:
                    technique_file.write("Technique ID: " + technique_id + "\n")
                    technique_file.write("Name: " + record["name"] + "\n\n")
                    technique_file.write("Description:\n" + record["description"] + "\n\n")
                    technique_file.write("Detection:\n" + record["x_mitre_detection"] + "\n")

                files_written = files_written + 1

            break

# Confirm how many files were written.
print("Files written:", files_written)
print("Expected:", len(labels))

# Check that every technique on the list got its file.
missing = []
for technique_id in labels:
    if not os.path.exists(os.path.join(data_folder, technique_id + ".txt")):
        missing.append(technique_id)

print("Missing:", missing)
```

### Setting up the Embedding model

```python
from llama_index.embeddings.openai import OpenAIEmbedding
from llama_index.core import VectorStoreIndex
from llama_index.core import Settings

# Turns a piece of text into a list of numbers.
# Texts with similar meaning get similar numbers.
Settings.embed_model = OpenAIEmbedding(model="text-embedding-3-large")
embed_model = OpenAIEmbedding(model="text-embedding-3-large")

# Each file is cut into pieces of about 512 tokens, with 20 tokens of
# overlap so no idea gets split in half.
Settings.chunk_size = 512
Settings.chunk_overlap = 20

print("Embedding model: text-embedding-3-large")
print("Chunk size:", Settings.chunk_size, "tokens")
print("Overlap:", Settings.chunk_overlap, "tokens")
```

### Getting all the files in the data folder

```python
def get_files(root_folder):
    files = []

    # Uses the same labels list defined further up.
    for foldername, subfolders, filenames in os.walk(root_folder):
        for filename in filenames:
            if filename.endswith(('.pdf', '.csv', '.txt')):
                if '_' not in filename:
                    if filename[:-4] in labels:
                        files.append(os.path.join(foldername, filename))
    return files


# Check what it picks up.
found = get_files(data_folder)
print("Files found:", len(found))
print("First three:", [os.path.basename(f) for f in found[:3]])
```

### Creating and saving the index

```python
from llama_index.core import SimpleDirectoryReader
from llama_index.core import StorageContext, load_index_from_storage
from llama_index.core.retrievers import VectorIndexRetriever
from llama_index.core.indices.postprocessor import SimilarityPostprocessor

import psutil

BATCH_SIZE = 200      # our 62 files fit in one batch

index = None

try:
    if not os.path.exists(PERSIST_DIR):
        # No saved index yet, so build one from the files.
        print("Creating folder:", PERSIST_DIR)
        os.mkdir(PERSIST_DIR)

        files = get_files(data_folder)
        print("Files to index:", len(files))

        if not files:
            raise ValueError("No files found in the data folder.")

        for i in range(0, len(files), BATCH_SIZE):
            batch_files = files[i:i + BATCH_SIZE]
            documents = SimpleDirectoryReader(input_files=batch_files).load_data()
            print("Loaded", len(documents), "documents.")

            if index is None:
                print("Making the embeddings. This is the part that costs money...")
                index = VectorStoreIndex.from_documents(documents=documents)
            else:
                for document in documents:
                    index.insert(document=document)

            print("Saving the index...")
            index.storage_context.persist(persist_dir=PERSIST_DIR)
            print("Saved.")
            print("Memory usage:", psutil.virtual_memory().percent, "%")

    else:
        # The index is already saved, so just load it.
        print("Loading the saved index from", PERSIST_DIR)
        storage_context = StorageContext.from_defaults(persist_dir=PERSIST_DIR)
        index = load_index_from_storage(storage_context=storage_context)
        print("Loaded.")

    if index is None:
        raise ValueError("Failed to load or create the index.")

    retriever = VectorIndexRetriever(index=index, similarity_top_k=20)
    postprocessor = SimilarityPostprocessor(similarity_cutoff=0.60)
    print("Done. Index and retriever are ready.")

except Exception as e:
    print("Caught an exception:", e)
    # Remove the folder if it was made but nothing got saved into it.
    if os.path.exists(PERSIST_DIR) and not os.listdir(PERSIST_DIR):
        os.rmdir(PERSIST_DIR)
```

```python
# Build a lookup so we can show the technique name, not just the ID.
technique_names = {}

for record in attack_data["objects"]:
    if record["type"] != "attack-pattern":
        continue
    for reference in record["external_references"]:
        if reference["source_name"] == "mitre-attack":
            technique_names[reference["external_id"]] = record["name"]
            break

print("Names loaded:", len(technique_names))
```
